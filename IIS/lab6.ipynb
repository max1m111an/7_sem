```python
import pandas as pd

def performance(x: int):
    if x <= 10:
        return -1
    elif x <= 12:
        return 0
    elif x <= 19:
        return 1

df = pd.read_csv("student-por.csv", delimiter=";")
display(df.describe())

df['performance'] = df['G3'].apply(lambda x: performance(x))

df.head(100)
display(df.describe())
```

```python
df.to_csv("studentpf.csv", index=False)
```
